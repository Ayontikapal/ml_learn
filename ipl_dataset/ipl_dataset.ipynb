```python
ipl_df['premium']=ipl_df['SOLD PRICE']-ipl_df['BASE PRICE'] 
ipl_df[['PLAYER NAME','SOLD PRICE','BASE PRICE','premium']].sort_values('premium', ascending=False)[0:5] #find the players that have the maximum premium value
```

```python
sp_age= ipl_df.groupby('AGE')['SOLD PRICE'].mean().reset_index() #this groups by age and results mean from sold price 
#results show that category 1 gets the highest sold price
print(sp_age)
```

```python
sp_ar= ipl_df.groupby(['AGE','PLAYING ROLE'])['SOLD PRICE'].mean().reset_index() #reset_index for get a dataframe and not series
print(sp_ar)
```

```python
sp_comp=sp_ar.merge(sp_age,on='AGE',how='outer')
sp_comp.rename(columns={'SOLD PRICE_x':'SOLD PRICE AGE ROLE','SOLD PRICE_y':'SOLD PRICE AGE'},inplace=True) #inplace avoid creating a new dataframe and makes changes in the old one itself
sp_comp
```

```python
sp_comp['CHANGE'] = sp_comp.apply(
    lambda rec: (rec["SOLD PRICE AGE ROLE"] - rec["SOLD PRICE AGE"])/ rec["SOLD PRICE AGE"],
    axis=1
)
sp_comp
```

```python
#filtering data for players who made more than 80 sixers
ipl_df[ipl_df['SIXERS']>80][['PLAYER NAME','SIXERS']].sort_values('SIXERS', ascending=False)
```
